## ROUND 3 · 디버깅 클리닉
오류가 나는 코드를 고치고, **왜 문제였는지**까지 말해보세요.

**A. 자료형 오류** — 힌트: `input`으로 받은 값은 기본적으로 글자
```python

```
**B. 조건문 오류** — 힌트: `:` 와 들여쓰기
```python
score = 60
if score >= 80
print("통과")
else:
print("조금 더 공부")
```
**C. 친절한 입력 처리** — 숫자가 아닐 때도 안 멈추게 바꿔라 (힌트: `try` / `except`)
```python
money = int(input("금액 입력: "))
print("입력한 금액:", money)
```

````python
# 🧑‍🏫 정답 A: 문자 + 숫자는 더할 수 없으므로 int()로 바꾼다
age = int(input("나이: "))
print(age + 1)
````

````python
# 🧑‍🏫 정답 B: if/else 끝에 ':' 추가 + 실행할 줄은 들여쓰기
score = 60
if score >= 80:
    print("통과")
else:
    print("조금 더 공부")
````

````python
# 🧑‍🏫 정답 C: 숫자가 아니면 ValueError → except에서 친절하게 안내
try:
    money = int(input("금액 입력: "))
    print("입력한 금액:", money)
except ValueError:
    print("숫자만 입력해주세요.")
````

## ROUND 4 · 코드 조립전
**MISSION A · 인사봇** — ① `print("안녕하세요,", name)` ② `name = input("이름을 입력하세요: ")` ③ `print("오늘은 파이썬을 배웁니다")`

**MISSION B · 금융 추천봇** — ① `product = input(...)` ② `print(tips.get(product, ...))` ③ `tips = {...}`

````python
# 🧑‍🏫 정답 A: ② → ① → ③
name = input("이름을 입력하세요: ")
print("안녕하세요,", name)
print("오늘은 파이썬을 배웁니다")
````

````python
# 🧑‍🏫 정답 B: ③ → ① → ② (보너스: 항목 하나 더 추가)
tips = {
    "예금": "안정적으로 돈을 모을 때 좋아요.",
    "적금": "매달 꾸준히 모을 때 좋아요.",
    "카드": "소비 관리가 중요해요.",
    "주식": "공부한 만큼만, 잃어도 괜찮은 돈으로 시작해요.",
}
product = input("궁금한 금융 상품은? ")
print(tips.get(product, "아직 준비되지 않았어요."))
````

## ROUND 5 · 미니 에이전트 제작 배틀
팀 수준에 맞는 난이도를 골라 **실제로 돌아가는 작은 프로그램**을 만드세요. 핵심은 "작아도 완성"입니다.

| 레벨 | 필수 요소 | 점수 |
|---|---|---|
| LEVEL 1 · 인사봇 | input 1개, 변수 1개, print 2개 이상 | 30 |
| LEVEL 2 · 추천봇 | input 1개, if/else 또는 dict.get(), 친절한 안내 문장 | 35 |
| LEVEL 3 · 상담 에이전트 | 함수 또는 class, 딕셔너리 또는 if, 질문/답변 2개 이상 | 40 |

추가 +10점: `try/except`, `for/list`, `random` 중 하나를 자연스럽게 넣기

````python
# LEVEL 1 · 인사봇 예시
name = input("이름이 뭐예요? ")
food = input("좋아하는 음식은? ")
print("반가워요,", name)
print(food, "좋아하는군요! 오늘 점심으로 어때요?")
````

````python
# LEVEL 2 · 추천봇 예시 (나이에 따라 안내 달라지기 + try/except 보너스)
try:
    age = int(input("나이를 입력하세요: "))
    if age >= 19:
        print("입출금 통장과 적금을 함께 만들어보세요.")
    else:
        print("보호자와 함께 청소년 통장을 만들어보세요.")
except ValueError:
    print("숫자로 입력해주세요.")
````

````python
# LEVEL 3 · 상담 에이전트 예시 (class + dict + random 보너스)
import random


class FinanceAgent:
    def __init__(self, name):
        self.name = name
        self.answers = {
            "저축": "용돈의 10%를 먼저 떼어 모아보세요.",
            "예금": "한 번에 맡기고 이자를 받는 상품이에요.",
            "적금": "매달 조금씩 넣어 목표 금액을 모아요.",
        }
        self.cheers = ["오늘도 화이팅!", "작은 습관이 큰 차이를 만들어요.", "벌써 반은 성공!"]

    def respond(self, question):
        for keyword, reply in self.answers.items():
            if keyword in question:
                return f"{self.name}: {reply} {random.choice(self.cheers)}"
        return f"{self.name}: 저축, 예금, 적금 중에 궁금한 걸 물어봐 주세요."


agent = FinanceAgent("파인 상담봇")
print(agent.respond(input("무엇이 궁금한가요? ")))
````
