## 6. 조건문 — `if` / `else`
`if` 다음 줄부터는 **들여쓰기**로 같은 묶음이라는 것을 보여줍니다. `if`가 맞지 않으면 `else`가 실행됩니다.

```python
score = 85
if score >= 80:
    print("좋아요")

spend = 12000
if spend <= 10000:
    print("예산 지킴")
else:
    print("지출 점검 필요")
```

## 7. 반복문 — `for` + `range`
`range(3)`은 0, 1, 2를 차례로 만들어 줍니다. 즉 3번 반복하라는 뜻입니다.

```python
print(list(range(3)))
print(1 in range(0, 3))

for i in range(3):
    print(i, "안녕하세요")
```

## 8. 리스트 `[ ]` — 여러 개를 한 번에 담는 상자
첫 번째 값은 `[0]`부터 시작합니다.

```python
foods = ["김밥", "라면", "떡볶이"]
banks = ["KB", "신한", "우리"]
products = ["예금", "적금", "카드"]

print(banks)
print(foods[0])
print(banks[1])
print(products[2])
```

### 자동화의 시작 — 리스트 + for

```python
banks = ["KB", "신한", "우리"]
for bank in banks:
    print(bank)

products = ["예금", "적금", "카드"]
for product in products:
    print(product, "추천")
```

## 9. 함수 — `def`로 자주 쓸 동작에 이름 붙이기
함수 이름 뒤에 `()`를 써야 실행됩니다. 괄호 안에 넣는 값(매개변수)에 따라 결과가 달라집니다.

```python
def hello():
    print("반갑습니다")

def greet(name):
    print("안녕하세요,", name)

def recommend(product):
    print(product, "설명을 시작할게요")

hello()
greet("지민")
greet("수아")
recommend("예금")
```

## 10. 딕셔너리 `{ }` — 이름표와 값이 짝지어진 상자
리스트가 순서대로 담는 상자라면, 딕셔너리는 이름표(key)와 값(value)을 짝지어 담습니다.

```python
student = {"name": "지민", "grade": 1}
account = {"bank": "KB", "money": 5000}

print(account)
print(student["name"])
```

### 딕셔너리로 아주 쉬운 챗봇 만들기
`get()`은 딕셔너리에서 값을 꺼내고, 없을 때 대신 보여줄 문장도 함께 넣을 수 있습니다.

```python
tips = {
    "예금": "안정적으로 돈을 모을 때 좋아요.",
    "적금": "매달 꾸준히 모을 때 좋아요.",
    "카드": "소비 관리가 중요해요.",
}

product = input("궁금한 금융 상품은? ")
print(tips.get(product, "아직 준비되지 않았어요."))
```

### 👥 팀 활동 — 딕셔너리에서 자주 쓰는 기능
`get()`, `keys()`, `values()`, `items()`, `update()`, `pop()`을 직접 사용하고 출력해보세요.

```python
student = {"이름": "알렉스", "나이": 20}

print(student.get("이름"))
print(list(student.keys()))
print(list(student.values()))
for key, value in student.items():
    print(key, "→", value)

student.update({"학교": "파인고"})
print(student)
student.pop("나이")
print(student)
```

### 👥 팀 활동 — `import json`?!
방금 만든 딕셔너리를 `json` 모듈로 바꿔봅시다. 왜 JSON 형식이 실무에서 더 좋을 수 있을까요?

> JSON은 **글자(문자열)** 이기 때문에 파일로 저장하거나, 다른 프로그램·다른 언어·웹 API와 주고받을 수 있습니다.

```python
import json

account = {"bank": "KB", "money": 5000, "owner": "지민"}

text = json.dumps(account, ensure_ascii=False)
print(type(text), text)

back = json.loads(text)
print(type(back), back["owner"])
```

## 11. 오류와 친해지기 — `try` / `except`
오류는 망했다는 뜻이 아니라, 어디를 고치면 되는지 알려주는 힌트입니다.
`input`으로 받은 값은 문자이므로 숫자 연산 전에 `int()`가 필요하고, 숫자가 아닌 값이 들어오면 `ValueError`가 납니다.

```python
try:
    money = int(input("금액 입력: "))
    print("입력한 금액:", money)
except ValueError:
    print("숫자만 입력해주세요.")
```

## 13. 클래스 — 값과 기능을 함께 묶는 설계도
`class`는 설계도, `self.name`은 그 설계도로 만든 대상(객체)이 가진 값입니다.

```python
class SavingBot:
    def __init__(self, name, product):
        self.name = name
        self.product = product

    def introduce(self):
        print("안녕하세요,", self.name, "입니다.")

    def recommend(self):
        print(self.product, "설명을 시작합니다.")


bot1 = SavingBot("금융 비서", "예금")
bot1.introduce()
bot1.recommend()

bot2 = SavingBot("절약 비서", "적금")
bot2.introduce()
```

## 14. 미니 에이전트 만들기
**입력받고 → 규칙 확인하고 → 답하는 프로그램.** 처음 단계의 에이전트는 생각보다 단순합니다.

### 가장 쉬운 버전 — 딕셔너리 + 입력

```python
answers = {
    "공부": "오늘 할 일을 10분짜리로 쪼개보세요.",
    "진로": "좋아하는 것과 잘하는 것을 같이 적어보세요.",
    "친구": "상대 이야기를 먼저 들어보세요.",
    "저축": "작게라도 목표 금액을 정해보세요.",
}

topic = input("고민을 입력하세요: ")
print(answers.get(topic, "조금 더 자세히 말해줄래요?"))
```

### 클래스 버전 — 도우미 안에 답변 기능 넣기
핵심은 “딕셔너리 응답 기능을 클래스 안으로 넣을 수도 있다”는 점입니다.
완성 버전은 터미널에서 `python mini_agent.py`로 실행해볼 수 있어요.

```python
class StudentAgent:
    def __init__(self):
        self.answers = {
            "공부": "오늘 할 일을 10분짜리로 나눠보세요.",
            "진로": "좋아하는 것과 잘하는 것을 같이 적어보세요.",
        }

    def respond(self, topic):
        return self.answers.get(topic, "조금 더 자세히 말해줄래요?")


agent = StudentAgent()
question = input("고민: ")
print(agent.respond(question))
```
